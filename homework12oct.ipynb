```python
student_marks = {
    "Physics": [86, 87, 43],
    "Maths": [67, 89, 78],
    "Chemistry": [98, 87, 76]
}

for k, v in student_marks.items():
    topper = v[0]
    roll_no = 1

    if v[1] > topper:
        topper = v[1]
        roll_no = 2

    if v[2] > topper:
        topper = v[2]
        roll_no = 3

    print(f"{k} Topper Roll No: {roll_no}, Marks: {topper}")
```
